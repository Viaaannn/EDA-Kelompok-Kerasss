| Kategori | Artinya |
|---|---|

```python
# TODO D1 — susun tabel penandaan Anda sendiri
# Buat dictionary bernama PENANDAAN berisi pasangan:
#     'nama_kolom': ('kategori', 'tindakan')
# Tindakan yang tersedia: 'hapus', 'hash', 'generalisasi', 'pertahankan'
# Kolom yang tidak Anda daftarkan dianggap 'bukan data pribadi' + 'pertahankan'.
#
# Lalu bangun DataFrame 'tanda' dengan kolom: nama_kolom, kategori, tindakan
# dan cetak jumlah kolom per kategori.
# Expected: pengenal langsung >= 5 kolom; data spesifik >= 3 kolom
```

```python
# TODO D1 — Susun tabel penandaan data pribadi

PENANDAAN = {}

# 1. PENGENAL LANGSUNG
# Data yang dapat digunakan untuk mengenali seseorang secara langsung
kolom_pengenal = [
    'id_peserta',
    'nama',
    'nama_lengkap',
    'nama_peserta',
    'nik',
    'no_telepon',
    'nomor_telepon',
    'no_hp',
    'email',
    'alamat'
]

for kolom in kolom_pengenal:
    if kolom in df.columns:
        tindakan = 'hash' if kolom == 'id_peserta' else 'hapus'
        PENANDAAN[kolom] = ('pengenal langsung', tindakan)


# 2. QUASI-IDENTIFIER
# Data yang dapat mengenali seseorang jika digabungkan
kolom_quasi = [
    'tanggal_lahir',
    'jenis_kelamin',
    'pendidikan_terakhir',
    'kelurahan',
    'kecamatan'
]

for kolom in kolom_quasi:
    if kolom in df.columns:
        PENANDAAN[kolom] = ('quasi-identifier', 'generalisasi')


# 3. DATA SPESIFIK
# Data pribadi yang membutuhkan perlindungan lebih
kolom_spesifik = [
    'penghasilan_pribadi_bulanan',
    'jumlah_anak',
    'status_pernikahan',
    'status_kesehatan'
]

for kolom in kolom_spesifik:
    if kolom in df.columns:
        PENANDAAN[kolom] = ('data spesifik', 'generalisasi')


# 4. MEMBUAT DATAFRAME PENANDAAN
tanda = pd.DataFrame([
    {
        'nama_kolom': kolom,
        'kategori': PENANDAAN.get(
            kolom, ('bukan data pribadi', 'pertahankan')
        )[0],
        'tindakan': PENANDAAN.get(
            kolom, ('bukan data pribadi', 'pertahankan')
        )[1]
    }
    for kolom in df.columns
])

display(tanda)

# Menghitung jumlah kolom berdasarkan kategori
print("\nJumlah kolom per kategori:")
print(tanda['kategori'].value_counts())

# Menampilkan jumlah pengenal langsung dan data spesifik
print("\nJumlah pengenal langsung:",
      (tanda['kategori'] == 'pengenal langsung').sum())

print("Jumlah data spesifik:",
      (tanda['kategori'] == 'data spesifik').sum())
```

```python
# TODO E2 — HASH
# Tulis fungsi samarkan(nilai, garam) yang mengembalikan 12 karakter pertama
# dari sha256(garam + nilai). Kembalikan np.nan bila nilainya kosong.
# Terapkan pada kolom yang Anda beri tindakan 'hash'.
# Expected: nilai asli dan hasil hash berbeda, dan nilai asli yang sama menghasilkan hash yang sama

# TODO E2 — HASH

GARAM = 'ganti-dengan-kalimat-rahasia-tim'

# Membuat fungsi untuk menyamarkan data menggunakan SHA-256
def samarkan(nilai, garam):
    if pd.isna(nilai):
        return np.nan

    teks = str(garam) + str(nilai)
    hasil_hash = hashlib.sha256(teks.encode()).hexdigest()

    return hasil_hash[:12]


# Mengambil kolom yang memiliki tindakan 'hash'
kolom_hash = tanda.loc[
    tanda['tindakan'] == 'hash',
    'nama_kolom'
].tolist()

# Menerapkan hash pada kolom yang telah ditentukan
for kolom in kolom_hash:
    aman[kolom] = aman[kolom].apply(
        lambda nilai: samarkan(nilai, GARAM)
    )


# Menampilkan hasil
print("Kolom yang di-hash:", kolom_hash)

for kolom in kolom_hash:
    print(f"\nContoh data asli dari kolom {kolom}:")
    print(df[kolom].head())

    print(f"Contoh hasil hash dari kolom {kolom}:")
    print(aman[kolom].head())

print("\nHasil DataFrame setelah proses hashing:")
display(aman.head())
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca
# a. Hash dua kali nilai yang sama, tunjukkan hasilnya identik
# b. Hash dua nilai berbeda, tunjukkan hasilnya berbeda

# TODO E3 — Membuktikan hash konsisten dan sulit dibaca

# A. Hash nilai yang sama sebanyak dua kali
nilai_sama = "peserta001"

hash_1 = samarkan(nilai_sama, GARAM)
hash_2 = samarkan(nilai_sama, GARAM)

print("A. Pengujian nilai yang sama")
print("Hash pertama :", hash_1)
print("Hash kedua   :", hash_2)
print("Hasil sama   :", hash_1 == hash_2)


# B. Hash dua nilai yang berbeda
nilai_1 = "peserta001"
nilai_2 = "peserta002"

hash_nilai_1 = samarkan(nilai_1, GARAM)
hash_nilai_2 = samarkan(nilai_2, GARAM)

print("\nB. Pengujian nilai yang berbeda")
print("Hash nilai pertama :", hash_nilai_1)
print("Hash nilai kedua   :", hash_nilai_2)
print("Hasil berbeda      :", hash_nilai_1 != hash_nilai_2)
```

```python
# TODO E4 — Generalisasi data umur dan wilayah

import pandas as pd
import numpy as np

# Menghitung umur berdasarkan tanggal 1 September 2025
tanggal_acuan = pd.Timestamp('2025-09-01')

aman['tanggal_lahir'] = pd.to_datetime(aman['tanggal_lahir'])

umur = (
    tanggal_acuan.year - aman['tanggal_lahir'].dt.year
    - (
        (tanggal_acuan.month < aman['tanggal_lahir'].dt.month) |
        (
            (tanggal_acuan.month == aman['tanggal_lahir'].dt.month) &
            (tanggal_acuan.day < aman['tanggal_lahir'].dt.day)
        )
    )
)

# Mengelompokkan umur
aman['kelompok_umur'] = pd.cut(
    umur,
    bins=[0, 24, 34, 44, 54, np.inf],
    labels=['<25', '25-34', '35-44', '45-54', '55+']
)

# Menghapus kolom yang terlalu spesifik
aman = aman.drop(columns=['tanggal_lahir', 'kelurahan'])

# Menyeragamkan penulisan kecamatan
aman['kecamatan'] = aman['kecamatan'].astype('string').str.strip().str.title()

# Menampilkan hasil
print("Distribusi kelompok umur:")
display(aman['kelompok_umur'].value_counts().sort_index())

print("\nJumlah kecamatan unik:")
print(aman['kecamatan'].nunique())

print("\nData setelah generalisasi:")
display(aman.head())
```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
# Kembalikan Series berisi: k (grup terkecil), grup berisi 1 orang, total grup


def cek_k(data, kolom):
    # Menghitung jumlah data pada setiap kombinasi kelompok
    jumlah_kelompok = data.groupby(
        kolom, dropna=False, observed=True
    ).size()

    # Menghitung nilai k terkecil
    nilai_k = jumlah_kelompok.min()

    # Menghitung jumlah kelompok yang hanya memiliki 1 anggota
    kelompok_tunggal = (jumlah_kelompok == 1).sum()

    # Menghitung total kelompok
    total_kelompok = len(jumlah_kelompok)

    return pd.Series({
        'k': nilai_k,
        'kelompok_beranggota_1': kelompok_tunggal,
        'total_kelompok': total_kelompok
    })

```

```python
# TODO F2 — ukur risiko SEBELUM penanganan
# Gunakan data asli df dengan kolom:
#   ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
# Expected: k = 1 dan SELURUH 620 baris berada di grup berisi satu orang

kolom_uji = [
    'kelurahan',
    'jenis_kelamin',
    'pendidikan_terakhir',
    'tanggal_lahir'
]

hasil_f2 = cek_k(df, kolom_uji)

print("Hasil pemeriksaan k-anonymity pada data awal:")
display(hasil_f2)
```

```python
# TODO F3 — ukur risiko SESUDAH generalisasi
# Gunakan 'aman' dengan kolom:
#   ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
# Expected: jumlah grup berisi satu orang turun banyak, tetapi k masih 1
```

```python

# TODO F3 — ukur risiko SESUDAH generalisasi

kolom_f3 = [
    'kecamatan',
    'jenis_kelamin',
    'pendidikan_terakhir',
    'kelompok_umur'
]

# Menghitung kombinasi dan jumlah kemunculan setiap grup
hasil_f3 = cek_k(aman, kolom_f3)

# Menampilkan hasil pengukuran risiko
print("Hasil Pengukuran Risiko Setelah Generalisasi:")
print(hasil_f3)

# Menghitung jumlah grup yang hanya berisi 1 orang
grup_satu = (
    aman.groupby(kolom_f3, observed=True)
    .size()
    .eq(1)
    .sum()
)

print("Jumlah grup berisi 1 orang:", grup_satu)
```

```python
# TODO F4 — TANTANGAN: capai k >= 5
# Generalisasikan lebih jauh sampai k >= 5, dengan MEMPERTAHANKAN sebanyak mungkin informasi.
# Anda boleh: memperlebar kelompok umur, menggabungkan kecamatan kecil menjadi 'Lainnya',
# atau mengeluarkan satu kolom dari daftar quasi-identifier.
# Catat setiap pilihan dan dampaknya terhadap k.
# Petunjuk: ada lebih dari satu jawaban benar. Yang dinilai adalah alasannya.

```

```python
# TODO F4 — TANTANGAN: capai k >= 5

# 1. Menggabungkan kelompok umur menjadi dua kategori
aman['kelompok_umur_f4'] = aman['kelompok_umur'].astype(str).replace({
    '<25': '<35',
    '25-34': '<35',
    '35-44': '35+',
    '45-54': '35+',
    '55+': '35+'
})

# 2. Menentukan quasi-identifier yang digunakan
kolom_f4 = ['kecamatan', 'kelompok_umur_f4']

# 3. Menghitung jumlah anggota setiap kelompok
grup_f4 = aman.groupby(kolom_f4, observed=True).size()

# 4. Menghitung nilai k
nilai_k = grup_f4.min()
grup_satu = (grup_f4 == 1).sum()
total_grup = len(grup_f4)

# 5. Menampilkan hasil
print("Hasil Generalisasi F4")
print("Nilai k:", nilai_k)
print("Jumlah grup berisi 1 orang:", grup_satu)
print("Total grup:", total_grup)

if nilai_k >= 5:
    print("TARGET TERCAPAI: k >= 5")
else:
    print("TARGET BELUM TERCAPAI")
```

Nilai k yang dicapai adalah 7, sehingga target k ≥ 5 berhasil terpenuhi. Informasi yang dikorbankan adalah detail kelompok umur, jenis kelamin, dan pendidikan terakhir untuk mengurangi risiko identifikasi ulang peserta._
